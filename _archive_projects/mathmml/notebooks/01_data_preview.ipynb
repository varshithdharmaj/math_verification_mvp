# Data Preview: GSM8K and Math500

This notebook explores the GSM8K and Math500 datasets.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path().absolute().parent))

from src.data.loaders import load_gsm8k, load_math500, create_step_dataset
import pandas as pd
from collections import Counter
```

```python
# Load GSM8K
gsm8k_train = load_gsm8k(".", "train", "main")
gsm8k_test = load_gsm8k(".", "test", "main")

print(f"GSM8K Train: {len(gsm8k_train)} examples")
print(f"GSM8K Test: {len(gsm8k_test)} examples")
```

```python
# Preview examples
for i in range(3):
    print(f"\nExample {i+1}:")
    print(f"Question: {gsm8k_train[i]['question']}")
    print(f"Answer: {gsm8k_train[i]['answer'][:200]}...")
```

```python
# Create step dataset
step_data = create_step_dataset(gsm8k_train[:100], inject_errors=True, error_ratio=0.5)

print(f"Total steps: {len(step_data)}")

# Count labels
labels = Counter([item['label'] for item in step_data])
print(f"\nLabel distribution:")
for label, count in labels.most_common():
    print(f"  {label}: {count}")
```
