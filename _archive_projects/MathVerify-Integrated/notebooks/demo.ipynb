# MathVerify-Integrated Demo

This notebook demonstrates the MathVerify-Integrated system for mathematical reasoning with verification.


```python
import sys
sys.path.append('..')

from src.pipeline import MathVerifyPipeline
```

```python
# Initialize pipeline
pipeline = MathVerifyPipeline()
```

```python
# Example problem
problem = "Solve for x: 2x + 3 = 7"

# Process through pipeline
result = pipeline.process_problem(problem)

# Display results
print(f"Problem: {result['problem']}")
print(f"\nFinal Answer: {result['final_answer']}")
print(f"Confidence: {result['confidence']}")
print(f"\nErrors: {result['errors']['total_errors']}")
```

```python
# View verified steps
for step in result['solution']['steps']:
    status = "✅" if step['is_valid'] else "❌"
    print(f"{status} Step {step['number']}: {step['content']}")
    if not step['is_valid']:
        print(f"   Error: {step.get('error_type', 'Unknown')}")
        print(f"   Suggestion: {step.get('correction', 'N/A')}")
```
