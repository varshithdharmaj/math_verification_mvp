# MathVerifyProject - Python API Demo

This notebook demonstrates how to use the MathVerifyProject Python API for mathematical reasoning verification.


## 1. Basic Setup

Import the necessary modules:


```python
from core_verification import MathVerifier

# Initialize verifier
verifier = MathVerifier()
print("✓ Verifier initialized")
```

## 2. Simple Verification

Verify a single answer (returns boolean):


```python
result = verifier.verify_answer(gold="1/2", prediction="0.5")
print(f"Result: {result}")
```

## 3. Detailed Verification

Get detailed information about the verification:


```python
result = verifier.verify_answer(
    gold="1/2", 
    prediction="0.5",
    return_details=True
)

print("Verification Result:")
print(f"  Valid: {result['valid']}")
print(f"  Gold: {result['gold']}")
print(f"  Prediction: {result['prediction']}")
if result.get('gold_parsed'):
    print(f"  Gold Parsed: {result['gold_parsed']}")
if result.get('pred_parsed'):
    print(f"  Prediction Parsed: {result['pred_parsed']}")
if result.get('error_type'):
    print(f"  Error Type: {result['error_type']}")
```
